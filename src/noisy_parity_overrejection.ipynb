# Noisy parity: high entropy, yet rejected by fair-coin correlation thresholds
Source $S_n=S_{n-1}\oplus\xi_n$ with $\Pr(\xi_n=1)=\frac12-\delta$ and $\delta=(2^\varepsilon-1)/2$, so $h_\infty=1-\varepsilon$.
Its lag-1 correlation sum is $\sum_n(-1)^{\xi_n}\approx2\delta N$, while the fair-coin rule allows only $\sqrt{2kN\ln N}$.
Here $\varepsilon=0.01$, $k=2$; margin $>1$ means rejection.

```python
import numpy as np
eps, k = 0.01, 2
delta = (2**eps - 1) / 2
rng = np.random.default_rng(20261012)
for N, T in [(10**5, 20), (10**6, 20), (10**7, 20), (10**8, 4)]:
    S = np.array([abs(N - 2 * int((rng.random(N) < 0.5 - delta).sum())) for _ in range(T)])  # lag-1 sum
    margin = S / np.sqrt(2 * k * N * np.log(N))
    print(N, (S / N).mean().round(5), margin.mean().round(3), f"rejected {(margin > 1).sum()}/{T}")
```

Check: for the real source the shortcut above is exact.

```python
xi = (rng.random(10**5 - 1) < 0.5 - delta).astype(int)
x = 1 - 2 * np.bitwise_xor.accumulate(np.r_[rng.integers(2), xi])   # S_n = S_{n-1} xor xi_n
assert (x[:-1] * x[1:]).sum() == (1 - 2 * xi).sum()
```
